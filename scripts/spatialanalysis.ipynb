### Geospatial analysis script to spatially compare perscribed burns to forest fires

##### Utilizing GeoPandas

```python
# Import Packages
from pathlib import Path
import pandas as pd
import geopandas as gpd
```

```python
# Create file paths
root_fldr = Path.cwd().parent
fire25_path = root_fldr / "data" / "raw" / "fire25_1.shp"
rxburn25_path = root_fldr / "data" / "raw" / "rxburn25_1.shp"
fire_rxburn_25_path = root_fldr / "data" / "processed" / "fire_rxburn_25.csv"
```

```python
# Import shapefiles as geodataframes and put both in the same coordinate reference system (CRS) for California
fire25 = gpd.read_file(fire25_path)
rxburn25 = gpd.read_file(rxburn25_path)
fire25 = fire25.to_crs(3310)
rxburn25 = rxburn25.to_crs(3310)
```

```python
# Convert the date fields for both geodataframes to datetime format
fire25['ALARM_DATE'] = pd.to_datetime(fire25['ALARM_DATE'], format='%Y,%m,%d')
rxburn25['START_DATE'] = pd.to_datetime(rxburn25['START_DATE'], format='%Y,%m,%d')
```

```python
# Spatially join the layers.
pairs = gpd.sjoin(
    fire25,
    rxburn25,
    how="inner",            # keep only fires that touch at least one burn
    predicate="intersects", # any overlap or shared edge counts as a match
    lsuffix="fire",
    rsuffix="rx",
)
```

```python
LOOKBACK_YEARS = 5 # Using 5 years but we can reset this threshold however we want

# Days between the prescribed burn and the fire
pairs["days_before_fire"] = (pairs["ALARM_DATE"] - pairs["START_DATE"]).dt.days

# Keep burns from 1 day up to 5 years before the fire
max_days = round(LOOKBACK_YEARS * 365.25) 
pairs = pairs[pairs["days_before_fire"].between(1, max_days)].copy()
```

```python
# Get overlap polygons with attributes from both layers
pairs = gpd.overlay(
    fire25[["INC_NUM", "ALARM_DATE", "geometry"]],
    rxburn25[["START_DATE", "geometry"]],
    how="intersection",
    keep_geom_type=True,
)

# Overlap area, then drop slivers
pairs["overlap_ha"] = pairs.geometry.area / 10_000
pairs = pairs[pairs["overlap_ha"] >= 1.0].copy()
```

```python
# Count qualifying burns and find the most recent one for each fire
history = pairs.groupby("INC_NUM").agg(
    rx_count=("START_DATE", "size"),
    last_rx_date=("START_DATE", "max"),
)

# Total treated area, merging overlapping burns so ground isn't counted twice
history["rx_overlap_ha"] = (
    pairs[["INC_NUM", "geometry"]].dissolve(by="INC_NUM").geometry.area / 10_000
)
```

```python
result = fire25.merge(history, how="left", left_on="INC_NUM", right_index=True)

# Fires with no qualifying burn get False / 0 instead of blanks
result["rx_within_window"] = result["rx_count"].notna()
result["rx_count"] = result["rx_count"].fillna(0).astype(int)
result["rx_overlap_ha"] = result["rx_overlap_ha"].fillna(0.0)

# Days from the most recent burn to the fire (blank if no burn)
result["days_since_rx"] = (result["ALARM_DATE"] - result["last_rx_date"]).dt.days
```

```python
result["fire_ha"] = result.geometry.area / 10_000
result["rx_overlap_pct"] = (
    result["rx_overlap_ha"] / result["fire_ha"] * 100
).clip(upper=100)
```

```python
# Query for fires with at least one burn within the 5-year window to compare to original data set
fires_with_rx = result.loc[result["rx_within_window"]].copy()
fires_with_rx.info()
```

```python
# Drop the geometry
result.drop(columns=["geometry"], inplace=True)
```

```python
# Save the result to a CSV file
result.to_csv(fire_rxburn_25_path, index=False)
```
