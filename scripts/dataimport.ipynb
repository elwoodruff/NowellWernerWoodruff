```python
from pathlib import Path
import pandas as pd
```

```python
#Create raw folder path
root_fldr = Path.cwd().parent
raw_fldr = root_fldr / "data" / "raw"
processed_fldr = root_fldr / "data" / "processed"

#Create raw csv file variables
fire25_csv = raw_fldr / "fire_25_TD.csv"
rxburn25_csv = raw_fldr / "rxburn_25_TD.csv"

#Import csv as data frames
fire25_raw = pd.read_csv(fire25_csv)
rxburn_raw = pd.read_csv(rxburn25_csv)
```

```python
#Report csv dimensions
print(fire25_raw.shape)
print(rxburn_raw.shape)
```

```python
#Clean imported data, filtering columns and rename
fire25 = fire25_raw.loc[:,['YEAR_', 'd_AGENCY', 'd_UNIT_ID', 'FIRE_NAME', 'INC_NUM', 'ALARM_DATE', 'CONT_DATE', 'GIS_ACRES', 'd_CAUSE', 'd_C_METHOD', 'd_OBJECTIVE', 'COMMENTS']]
rxburn25 = rxburn_raw.loc[:,['YEAR_', 'd_AGENCY', 'd_UNIT_ID', 'TREATMENT_ID', 'd_TREATMENT_TYPE', 'START_DATE', 'END_DATE', 'GIS_ACRES', 'd_PRE_CON_CLASS', 'd_POST_CON_CLASS', 'd_RX_CONSUM']]

#Convert Year to Integer data type
fire25['YEAR_'] = fire25['YEAR_'].astype('Int64') #pd.to_datetime(fire25['YEAR_'],format='%Y')
rxburn25['YEAR_'] = rxburn25['YEAR_'].astype('Int64')#pd.to_datetime(rxburn25['YEAR_'],format='%Y')

```

```python
#Parse dates
fire25['ALARM_DATE'] = pd.to_datetime(fire25['ALARM_DATE'],format='%m/%d/%Y %H:%M:%S')
fire25['CONT_DATE'] = pd.to_datetime(fire25['CONT_DATE'],format='%m/%d/%Y %H:%M:%S')

rxburn25['START_DATE'] = pd.to_datetime(rxburn25['START_DATE'],format='%m/%d/%Y %H:%M:%S')
rxburn25['END_DATE'] = pd.to_datetime(rxburn25['END_DATE'],format='%m/%d/%Y %H:%M:%S')
```

```python
#Rename columns
fire25 = fire25.rename(
    columns={
        'YEAR_': 'YEAR',
        'd_AGENCY': 'AGENCY',
        'd_UNIT_ID': 'UNIT_ID',
        'd_CAUSE': 'CAUSE',
        'd_OBJECTIVE': 'OBJECTIVE',
        'd_C_METHOD': 'C_METHOD'
    }
)

rxburn25 = rxburn25.rename(
    columns={
        'YEAR_': 'YEAR',
        'd_AGENCY': 'AGENCY',
        'd_UNIT_ID': 'UNIT_ID',
        'd_TREATMENT_TYPE': 'TREATMENT_TYPE',
        'd_PRE_CON_CLASS': 'PRE_CON_CLASS',
        'd_POST_CON_CLASS': 'POST_CON_CLASS',
        'd_RX_CONSUM': 'RX_CONSUM'
    }
)
```

```python
#Report fire null values
print(fire25.isna().sum())
```

```python
#Report rxburn null values
print(rxburn25.isna().sum())

```

```python
# Filter for data only after the year 2000
fire25_2000 = fire25.loc[fire25['YEAR'] >= 2000]
rxburn25_2000 = rxburn25.loc[rxburn25['YEAR'] >= 2000]
```

```python
# Report info on filtered fire data
print(fire25_2000.info())
# Report info on filtered rxburn data
print(rxburn25_2000.info())
```

```python
# Save filtered and cleaned data to processed folder
fire25_2000.to_csv(processed_fldr / 'fire25_2000.csv', index=False)
rxburn25_2000.to_csv(processed_fldr / 'rxburn25_2000.csv', index=False)
```
